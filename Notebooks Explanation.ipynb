# (1) 🏥 Patient Heart Rate Simulator
**Description:** Generates synthetic, high-frequency heart rate data for a 26-year-old male using a State-Switching Autoregressive model.

---
### 🛠️ Configuration
| Parameter | Description |
| :--- | :--- |
| **Catalog/Schema** | `production_catalog` / `patient_data` |
| **Output Volume** | `/Volumes/production_catalog/patient_data/raw_jsons/` |
| **Model Type** | Markov Chain + AR(1) Mean Reversion |

### 📈 Logic Overview
1. **State Transition:** Probabilistic switch between REST, EXERCISE, and CRISIS.
2. **Value Drift:** BPM moves toward state targets ($T_s$) with Gaussian noise.
3. **Storage:** Writes individual JSON files to the specified Volume.

# (2) 🧹 Volume Maintenance: Daily Data Retention
**Description:** Automated task that scans the raw JSON volume and removes files exceeding the 24-hour retention threshold to optimize storage and maintain performance.

---

### ⚙️ Policy Settings
* **Target Path:** `/Volumes/production_catalog/patient_data/raw_jsons/`
* **Retention Window:** `1 Day` (24 Hours)
* **Execution:** Scheduled (Daily Job)
* **Logic:** Files with `modificationTime` < `(now - 24h)` are permanently removed.
* **Error Handling:** Configured with Email Notifications to alert engineering teams immediately upon task failure or volume access issues.

### ⚠️ Safety Warnings
> [!WARNING]
> This notebook performs **permanent deletions** using `dbutils.fs.rm`. 
> Data deleted from Volumes cannot be recovered. Ensure the `VOLUME_PATH` is correctly mapped to the "Raw" landing zone.

### 📋 Monitoring
* Check the **Job Run History** in the Workflows tab to verify the count of deleted vs. retained files for each daily run.
